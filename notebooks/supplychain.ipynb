```python
print("Duplicates:", df.duplicated().sum())
print("Missing values:\n", df.isna().sum()[df.isna().sum() > 0])

# Confirm these column pairs are identical before dropping one of each
print((df["Benefit per order"] == df["Order Profit Per Order"]).all())
print((df["Sales per customer"] == df["Order Item Total"]).all())

# Financial validation
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Negative profit rows:", (df["Benefit per order"] < 0).sum())
print("Zero profit rows:", (df["Benefit per order"] == 0).sum())
print(df["Order Status"].value_counts())
```

```python
# 1. Drop duplicate columns (keep one of each pair)
df = df.drop(columns=["Order Profit Per Order", "Sales per customer"])

# 2. Fill the few missing last names, build a full name
df["Customer Lname"] = df["Customer Lname"].fillna("")
df["Customer Name"] = (df["Customer Fname"] + " " + df["Customer Lname"]).str.strip()

# 3. Remove orders that aren't real revenue
df = df[~df["Order Status"].isin(["CANCELED", "SUSPECTED_FRAUD"])].copy()

# 4. Remove zero-profit rows (as the brief says); KEEP negative profits
df = df[df["Benefit per order"] != 0].copy()

# 5. Drop columns we won't use
df = df.drop(columns=["Customer Street", "Customer Zipcode"])

print("Rows before:", len(df_raw), "| after:", len(df))
```

```python
# Net sales after discount (this column already exists)
df["Net Sales"] = df["Order Item Total"]

# Row-level margin %
df["Margin %"] = df["Benefit per order"] / df["Net Sales"] * 100

# Discount bands (to study margin erosion later)
df["Discount Band"] = pd.cut(
    df["Order Item Discount Rate"],
    bins=[-0.01, 0, 0.05, 0.10, 0.15, 0.20, 0.25],
    labels=["0%", "0-5%", "5-10%", "10-15%", "15-20%", "20-25%"]
)

# Flag discounted vs non-discounted orders
df["Discounted"] = np.where(df["Order Item Discount"] > 0, "Yes", "No")
```

```python
total_revenue = df["Sales"].sum()
total_net     = df["Net Sales"].sum()
total_profit  = df["Benefit per order"].sum()
margin        = total_profit / total_net * 100

print(f"Gross Revenue: {total_revenue:,.0f}")
print(f"Net Revenue:   {total_net:,.0f}")
print(f"Total Profit:  {total_profit:,.0f}")
print(f"Profit Margin: {margin:.2f}%")
```

```python
# orders where profit is negative
loss_orders = df[df["Benefit per order"] < 0]

print("Number of loss orders:", len(loss_orders))
print("Total money lost:", loss_orders["Benefit per order"].sum())
```

```python
# add up sales and profit for each segment
segment = df.groupby("Customer Segment")[["Net Sales", "Benefit per order"]].sum()

# margin = profit / sales * 100
segment["Margin %"] = segment["Benefit per order"] / segment["Net Sales"] * 100

print(segment)
```

```python
market = df.groupby("Market")[["Net Sales", "Benefit per order"]].sum()
market["Margin %"] = market["Benefit per order"] / market["Net Sales"] * 100

print(market)
```

```python
category = df.groupby("Category Name")[["Net Sales", "Benefit per order"]].sum()
category["Margin %"] = category["Benefit per order"] / category["Net Sales"] * 100

# lowest margin first
category = category.sort_values("Margin %")
print(category.head(10))
```

```python
product = df.groupby("Product Name")[["Net Sales", "Benefit per order"]].sum()
product["Margin %"] = product["Benefit per order"] / product["Net Sales"] * 100

print(product.sort_values("Net Sales", ascending=False).head(10))
```

```python
# overall company margin
overall_margin = df["Benefit per order"].sum() / df["Net Sales"].sum() * 100

# the sales level that separates the top 25% of products
high_sales_limit = product["Net Sales"].quantile(0.75)

# high sales AND margin below the company average
problem_products = product[(product["Net Sales"] >= high_sales_limit) & (product["Margin %"] < overall_margin)]

print("Overall margin:", overall_margin)
print(problem_products)
```

```python
# total profit for each customer, highest first
customer_profit = df.groupby("Customer Id")["Benefit per order"].sum()
customer_profit = customer_profit.sort_values(ascending=False)

# top 20% of customers
top_20_count = int(len(customer_profit) * 0.20)
top_20_profit = customer_profit.head(top_20_count).sum()

print("Share of profit from top 20% customers:", top_20_profit / customer_profit.sum() * 100)
```

```python
import matplotlib.pyplot as plt

plt.scatter(product["Net Sales"], product["Margin %"])
plt.axhline(overall_margin, color="red")
plt.xlabel("Net Sales")
plt.ylabel("Margin %")
plt.title("Product Sales vs Margin")
plt.show()
```

```python
# add up sales and profit for each customer
customer = df.groupby("Customer Id")[["Net Sales", "Benefit per order"]].sum()

# rename the profit column to something shorter
customer = customer.rename(columns={"Benefit per order": "Profit"})

# count how many orders each customer made
customer["Orders"] = df.groupby("Customer Id")["Benefit per order"].count()

# margin of each customer
customer["Margin %"] = customer["Profit"] / customer["Net Sales"] * 100

print(customer.shape)
print(customer.head())
```

```python
sorted_customers = customer.sort_values("Profit", ascending=False)

print("Top 10 customers by profit:")
print(sorted_customers.head(10))

print("Bottom 10 customers by profit:")
print(sorted_customers.tail(10))
```

```python
loss_customers = customer[customer["Profit"] < 0]

print("Loss-making customers:", len(loss_customers))
print("Share of all customers:", len(loss_customers) / len(customer) * 100)
print("Total loss from them:", loss_customers["Profit"].sum())
```

```python
# only look at profitable customers to set the limits
profitable = customer[customer["Profit"] > 0]

low_limit = profitable["Profit"].quantile(0.25)
high_limit = profitable["Profit"].quantile(0.75)

# put each customer in a tier
customer["Tier"] = pd.cut(
    customer["Profit"],
    bins=[-float("inf"), 0, low_limit, high_limit, float("inf")],
    labels=["Loss-making", "Low value", "Medium value", "High value"]
)

print(customer["Tier"].value_counts())
```

```python
tier_summary = customer.groupby("Tier", observed=True)[["Net Sales", "Profit"]].sum()

tier_summary["Customers"] = customer["Tier"].value_counts()
tier_summary["Margin %"] = tier_summary["Profit"] / tier_summary["Net Sales"] * 100
tier_summary["Profit Share %"] = tier_summary["Profit"] / tier_summary["Profit"].sum() * 100

print(tier_summary)
```

```python
# profit and number of unique customers in each segment
seg_profit = df.groupby("Customer Segment")["Benefit per order"].sum()
seg_customers = df.groupby("Customer Segment")["Customer Id"].nunique()

# Customer Value Index = profit per customer
value_index = seg_profit / seg_customers

print(value_index)
```

```python
# group by Yes/No and add up sales and profit
disc = df.groupby("Discounted")[["Net Sales", "Benefit per order"]].sum()

disc["Orders"] = df["Discounted"].value_counts()
disc["Margin %"] = disc["Benefit per order"] / disc["Net Sales"] * 100

print(disc)
```

```python
band = df.groupby("Discount Band", observed=True)[["Net Sales", "Benefit per order"]].sum()

band["Orders"] = df["Discount Band"].value_counts()
band["Margin %"] = band["Benefit per order"] / band["Net Sales"] * 100

print(band)
```

```python
# True/False column: is this order a loss?
df["Loss Order"] = df["Benefit per order"] < 0

# the average of True/False gives the share of losses
loss_share = df.groupby("Discount Band", observed=True)["Loss Order"].mean() * 100

print(loss_share)
```

```python
rate = df.groupby("Order Item Discount Rate")[["Net Sales", "Benefit per order"]].sum()
rate["Margin %"] = rate["Benefit per order"] / rate["Net Sales"] * 100

print(rate)
```

```python
total_discount = df["Order Item Discount"].sum()
total_gross_sales = df["Sales"].sum()
total_profit = df["Benefit per order"].sum()

# actual margin (with discounts)
actual_margin = total_profit / df["Net Sales"].sum() * 100

# margin if no discount had been given (assuming costs stay the same)
margin_without_discount = (total_profit + total_discount) / total_gross_sales * 100

print("Total discount given:", total_discount)
print("Discount as % of gross sales:", total_discount / total_gross_sales * 100)
print("Actual margin:", actual_margin)
print("Margin without discounts:", margin_without_discount)
print("Margin lost due to discounts:", margin_without_discount - actual_margin)
```

```python
import matplotlib.pyplot as plt

# chart 1: margin by discount band
plt.bar(band.index.astype(str), band["Margin %"])
plt.xlabel("Discount Band")
plt.ylabel("Margin %")
plt.title("Profit Margin by Discount Band")
plt.show()

# chart 2: loss-making order share by band
plt.bar(loss_share.index.astype(str), loss_share.values)
plt.xlabel("Discount Band")
plt.ylabel("Loss Orders %")
plt.title("Share of Loss-Making Orders by Discount Band")
plt.show()
```
